#### Machine learning 
is a subfield of artificial intelligence (AI) that focuses on enabling computers to learn from data without being explicitly programmed.  
* Without Explicit Programming  
* Making Predictions or Decisions  



#### Neural Networks:  
 type of machine learning (ML) process, that uses interconnected nodes or neurons in a LAYERED structure that resembles the human brain.    

Are computational models with interconnected nodes, forming layers.  
Can be simple or complex, with varying numbers of layers.  
A broader term, encompassing various network architectures.  
#### Deep Learning:  

Deep learning is a subset of machine learning that uses MULTILAYERD neural networks, to simulate the complex decision-making power of the human brain  


Is a subfield of machine learning that uses neural networks with many (deep) layers.  
Focuses on learning complex patterns from data through these deep architectures.  
Is a specific application of neural networks, utilizing their depth for advanced tasks.  
Essentially, deep learning is a specialized application of neural networks, where the "depth" of the network is the key differentiating  factor.   

###### NOTE: All deep learning models are neural networks. BUT not all neural networks are considered deep learning  

##### Key points:

Depth: The primary difference is the number of layers. Deep learning involves neural networks with multiple hidden layers.     
Complexity: Deep learning models are generally more complex and require more computational resources than traditional neural networks.     
Applications: Deep learning excels in tasks where traditional machine learning and neural networks struggle, such as image recognition, natural language processing, and speech recognition
