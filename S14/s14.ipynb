##### Supervised Machine learning: 

ALGORITHM: in ML describe the process of learning from data.  
An “algorithm” in machine learning is a procedure that is run on data to create a machine learning “model    
* Classification:  discrette lables:   KNN, LOR ,  Preceptrons, (MLPs)(Neural network), Support Vector Machines (SVMs) 
* Regression:  continous labels:     SLR, MLR  , multilayer perceptrons (MLPs)(Neural network), Support Vector Machines (SVMs)

NOTE: In regression the aim is to predict the y-value (the true label) based on the x-value.  
NOTE: The MSE measures the average squared difference between the predicted y-values (ŷ) and the actual y-values (y).    
              * Simple(one feature, one label) : linear, non linear   

                            ** Linear: There is direct relationship between X and Y (x = independant, y = dependant)  
                            ** Nonlinear: There is NO direct relationship between X and Y ((x = independant, y = dependant))
 
              * Multiple(multiple feature, one label): linear, non linear
