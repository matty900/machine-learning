Overfitting: Model works good on train data but not test data - it memorize not generalize  
              Overfitting occurs when a model learns the training data too well and fails to generalize to new, unseen data.  

Techniques to prevent Overfitting: 
1. Batch normalization: it's on top of neural network to normlize acitivation layer by layer - normalization on batch no whole layer 
2. Data Augmentation: to increase the Size and Diversity of the training data - add noise to data  
              artificially increase the size of a training dataset by creating modified versions of existing data.  
3. Learning rate: A hyperparameter that controls how much the model updates its weights in response to the loss gradient.  
• Too high → Model diverges  
• Too low → Slow convergence  
• Just right → Fast & stable training  
NOTE: decay in code: gradually slowing the learning rate

Data Augmentation purpose:   
1. Increase Dataset Size   
2. Improve Model Generalization  
3. Enhance Model Robustness  

How it Works:  
1. Image Augmentation: Rotation   Flipping (horizontal, vertical)    Cropping    Scaling (zooming in or out)    Brightness and contrast adjustments   Adding noise  
2. Audio Augmentation: Adding background noise    Changing pitch or speed    Time shifting        
3. Text Augmentation: Synonym replacement    Word or sentence shuffling    Random insertion or deletion of words
